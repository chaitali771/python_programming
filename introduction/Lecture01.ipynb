```python
train_name = input("Enter train name: ")

total_seats = 100
booked_seats = 60

vacant_seats = total_seats - booked_seats

print("Train Name:", train_name)
print("Vacant Seats:", vacant_seats)

booking = input("Do you want to book a ticket? (yes/no): ")

if booking == "yes":
    tickets = int(input("How many tickets do you want? "))

    if tickets <= vacant_seats:
        booked_seats = booked_seats + tickets
        vacant_seats = total_seats - booked_seats

        print("Booking Successful")
        print("Number of Bookings:", tickets)
        print("Vacant Seats:", vacant_seats)
    else:
        print("Sorry, tickets are not available.")

else:
    print("Booking Cancelled")
```
